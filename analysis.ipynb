```python
import pandas as pd
import json
```

```python
data = []
for split in ("train", "dev"):
    data.append(pd.read_json(f"../lexical_datasets/xl-lexeme.{split}.json"))
data = pd.concat(data).reset_index(drop=True)
```

```python
data.to_json("data/xl-lexeme.json", indent=2, orient="records", force_ascii=False)
```

```python
test_lemmas = set(pd.read_json("data/mcl-wic.test.json").lemma.str.lower().values)
```

```python
data = pd.read_json("predictions_sense_wic_xl-lexeme.json")
```

```python
columns = ["reasonings", "answers", "votes"]
data = data.explode(columns, ignore_index=True).sample(frac=1.0, random_state=42)
```

```python
data = data[~data.lemma.str.lower().isin(test_lemmas)]
```

```python
data
```

```python
for idx, item in data.sample(10).iterrows():
    print("-"*80)
    print("s1:", item['sentence1'])
    print("s2:", item['sentence2'])
    print("<think>")
    print(item['reasonings'])
    print("</think>")
    print(json.loads(item['answers']))
    print("GT:", item['label'])
    print("-"*80)
```

```python
data = data[~data.pos.isna()]
```

```python
data[data.label.astype(bool) == data.votes].to_json("data/lexeme.correct.json", orient="records", indent=2)
```
